```python
#Read a file line by line and store it in a list.
lines = []
with open("sample.txt","r") as file:
    for line in file:
        lines.append(line.strip())
print(lines)
```

```python
%%writefile input.txt
Red
Black
Orange
violet
Pink
```

```python
#Copy odd lines of one file to another.
with open("input.txt","r") as src:
    lines=src.readlines()
with open("odd_lines.txt","w") as dest:
    for i in range(len(lines)):
        if i % 2 ==0:
            dest.write(lines[i])
print("Odd lines copied successfully!")
```

```python
#Read each row from a csv file and print a list of strings.
import csv
with open("student.csv","r") as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
#Write a python dictionary to a csv file,read it,and display its content.
import csv
studentdetail=[
    {'Name':'Anu','Age':20,'Department':'MCA'},
    {'Name':'Rahul','Age':21,'Department':'MBA'},
    {'Name':'Arun','Age':19,'Department':'BCA'}]
with open("studentdetail.csv","w",newline="") as file:
    fields=['Name','Age','Department']
    writer=csv.DictWriter(file,fieldnames=fields)
    writer.writeheader()
    writer.writerows(studentdetail)
print("Dictionary written successfully!")
```
