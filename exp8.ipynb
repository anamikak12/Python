```python
#accepting 5 students names and rollno,and these map into a dictionary
rollno=input("enter 5 students rollno:").split(",")
name=input("enter 5 students name:").split(",")
my_dict=dict(zip(rollno,name))
print("orginal dictionary is",my_dict)
print("dictionary in ascending order:",dict(sorted(my_dict.items())))
print("dictionary in descending order:",dict(sorted(my_dict.items(),reverse=True)))
```

```python
#merge 2 dict and display
d1=dict(zip(input("enter 3 roll number:").split(","),input("enter 3 names:").split(",")))
d2=dict(zip(input("enter 2 roll number:").split(","),input("enter 2 names:").split(",")))
print("dictionary1 :",d1)
print("dictionary 2:",d2)
d1.update(d2)
print("merged dictionary:",d1)
```
