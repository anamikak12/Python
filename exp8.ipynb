```python
#accepting 5 students names and rollno,and these map into a dictionary
name=input("enter 5 students name:").split(",")
rollno=input("enter 5 students rollno:").split(",")
my_dict=dict(zip(name,rollno))
asc=dict(sorted(my_dict.items()))
des=dict(sorted(my_dict.items(),reverse=True))
print("dictionary in ascending order:",asc)
print("dictionary in descending order:",des)
```
