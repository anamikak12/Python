```python
#changea given string to a new string
string=input("enter a string:")
new_string=string[-1]+string[1:-1]+string[0]
print("new string:",new_string)
```

```python
#add the first and last two characters from the string
s=input("enter a string:")
new_string=s[:2]+s[-2:]
print("new string:",new_string)
```

```python
#changing the first letter of each string
s1=input("enter first string:")
s2=input("enter second string:")
st1=s2[0]+s1[1::1]
st2=s1[0]+s2[1::1]
print(st1 +" "+ st2)
```

```python
#replacing the first letter with $
s=input("enter a string:")
f=s[0]
result = f + s[1:].replace(f,"$")
print("result :",result)
```
