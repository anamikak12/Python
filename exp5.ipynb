```python
#third element and third element from last of a tuple
t=tuple(input("enter the elements separated by comma").split(","))
print("3rd element:",t[2])
print("3rd element from last:",t[-3])
```

```python
#find avg of all items in that tuple
n=input("enter the numbers:")
tu=tuple(map(int,n.split(",")))
print("tuple is:",tu)
print("average is:",sum(tu)/len(tu))
```
