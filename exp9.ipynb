```python
# add "ing" or "ly" to the end of the string
s=input("enter a string:");
if s[-3:]=="ing":
   s1=s+"ly"
else:
   s1=s+"ing"
print("orginal string is:",s)
print("new string is:",s1)
```

```python
#check its paliindrome or not
s=input("enter a string:")
print("orginal string is:",s)
if s==s[::-1]:
    print("it is a palindrome")
else:
    print("it is not a palindrome")
```
