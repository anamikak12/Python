```python
#2.area of the circle by inputingradius from math import pi
import math
r=int(input("enter the radius of a circle="))
area=round(math.pi*r*r,2)
print("area of a circle:"f"{area:.2f}")
```

```python
#3.biggest of three input number using max()
n1=int(input("enter first number:"))
n2=int(input("enter the second number:"))
n3=int(input("enter third number:"))
max=max(n1,n2,n3)
print("maximum no:",max)
```
