```python
#Convert Celsius to Fahrenheit
c=float(input("Enter the temperature in Celsius:"))
f=(c*9/5)+32
print("The corresponding Fahrenheit:",f)
```

```python
#Program to convert kilometer into miles
km=float(input("Enter the distance in km:"))
m=km*0.62137
print("The corresponding miles:",f"{m:.2f}")
```
