```python
#compare 2 rectangle by their area
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)

    def compare(self, other):
        if self.area() > other.area():
            print("First rectangle has a larger area.")
        elif self.area() < other.area():
            print("Second rectangle has a larger area.")
        else:
            print("Both rectangles have equal area.")
            
l1 = float(input("Enter length of first rectangle: "))
b1 = float(input("Enter breadth of first rectangle: "))
r1 = Rectangle(l1, b1)
print("\nFirst Rectangle")
print("Area:", r1.area())
print("Perimeter:", r1.perimeter())

l2 = float(input("Enter length of second rectangle: "))
b2 = float(input("Enter breadth of second rectangle: "))
r2 = Rectangle(l2, b2)
print("\nSecond Rectangle")
print("Area:", r2.area())
print("Perimeter:", r2.perimeter())

r1.compare(r2)
```

```python
#bank account
class Bank:
    def __init__(self, acc_no, name, acc_type, balance):
        self.acc_no = acc_no
        self.name = name
        self.acc_type = acc_type
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Amount deposited successfully")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                print("Amount withdrawn successfully")
            else:
                print("Insufficient balance")
        else:
            print("Invalid withdrawal amount")

    def display(self):
        print("\nAccount Number:", self.acc_no)
        print("Name:", self.name)
        print("Account Type:", self.acc_type)
        print("Balance:", self.balance)

acc_no = int(input("Enter account number: "))
name = input("Enter name: ")
acc_type = input("Enter account type: ")
balance = float(input("Enter initial balance: "))

b = Bank(acc_no, name, acc_type, balance)

while True:
    print("\n----- BANK MENU -----")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Display")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = float(input("Enter deposit amount: "))
        b.deposit(amount)

    elif choice == 2:
        amount = float(input("Enter withdrawal amount: "))
        b.withdraw(amount)

    elif choice == 3:
        b.display()

    elif choice == 4:
        print("Thank you for using our bank!")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
#information about book publications
class Publisher:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Publisher Name:", self.name)


class Book(Publisher):
    def __init__(self, name, title, author):
        super().__init__(name)        
        self.author = author
        self.title = title
    def display(self):
        print("Publisher Name:", self.name)
        print("Book Title:", self.title)
        print("Author:", self.author)


class Python(Book):
    def __init__(self, name, title, author, price, no_of_pages):
        super().__init__(name, title, author)
        self.price = price
        self.no_of_pages = no_of_pages

    def display(self):  
        super().display()
        print("Price:", self.price)
        print("Number of Pages:", self.no_of_pages)


name = input("Enter publisher name: ")
title = input("Enter book title: ")
author = input("Enter author name: ")
price = float(input("Enter book price: "))
no_of_pages = int(input("Enter number of pages: "))

p = Python(name, title, author, price, no_of_pages)

print("\n--- Python Book Details ---")
p.display()
```

```python
#Operator Overloading to Compare the Area of Two Rectangles
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def __lt__(self, other):
        return self.area() < other.area()

l1 = int(input("Enter length of first rectangle: "))
w1 = int(input("Enter width of first rectangle: "))
r1 = Rectangle(l1, w1)
print("Area of first rectangle:", r1.area())

l2 = int(input("Enter length of second rectangle: "))
w2 = int(input("Enter width of second rectangle: "))
r2 = Rectangle(l2, w2)
print("Area of second rectangle:", r2.area())

if r1 < r2:
    print("Second rectangle has greater area")
else:
    print("First rectangle has greater or equal area")
```

```python
#Operator Overloading to Add Two Time Objects
class Time:
    def __init__(self, hour, minute, second):
        self.__hour = hour
        self.__minute = minute
        self.__second = second

    def __add__(self, other):
        s = self.__second + other.__second
        m = self.__minute + other.__minute
        h = self.__hour + other.__hour

        m = m + s // 60
        s = s % 60

        h = h + m // 60
        m = m % 60

        return Time(h, m, s)

    def display(self):
        print("Time:", self.__hour, ":", self.__minute, ":", self.__second)
        
h1 = int(input("Enter hour of first time: "))
m1 = int(input("Enter minute of first time: "))
s1 = int(input("Enter second of first time: "))

h2 = int(input("Enter hour of second time: "))
m2 = int(input("Enter minute of second time: "))
s2 = int(input("Enter second of second time: "))

t1 = Time(h1, m1, s1)
t2 = Time(h2, m2, s2)

t3 = t1 + t2
print("First time:")
t1.display()

print("Second time:")
t2.display()

print("Sum of two times:")
t3.display()
```
